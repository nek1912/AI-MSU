```python
from pathlib import Path
import pandas as pd

PROJECT_DIR = Path(r"D:\AI-MSU")
RAW_DIR = PROJECT_DIR / "data" / "raw"
PROCESSED_DIR = PROJECT_DIR / "data" / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

def find_file(filename):
    files = list(RAW_DIR.rglob(filename))
    if len(files) == 0:
        raise FileNotFoundError(filename + " not found")
    return files[0]

logon_path = find_file("logon.csv")
sample = pd.read_csv(logon_path, nrows=10000)
print("Rows and columns before cleaning:", sample.shape)
display(sample.head())
```

```python
print("Missing values in each column:")
display(sample.isnull().sum().sort_values(ascending=False))

print("Duplicate rows:", sample.duplicated().sum())
```

```python
cleaned = sample.drop_duplicates().copy()

if "date" in cleaned.columns:
    cleaned["date"] = pd.to_datetime(cleaned["date"], errors="coerce")

if "user" in cleaned.columns:
    cleaned = cleaned.dropna(subset=["user"])

print("Rows and columns after cleaning:", cleaned.shape)
print("Missing values after cleaning:")
display(cleaned.isnull().sum())
display(cleaned.head())
```

```python
cleaned.to_csv(PROCESSED_DIR / "logon_cleaning_sample.csv", index=False)
print("Saved a cleaned sample. The full raw file has not been changed.")
```
